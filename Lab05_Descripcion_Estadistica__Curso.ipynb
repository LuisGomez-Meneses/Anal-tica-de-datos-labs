```python
import pandas as pd
df = pd.read_csv("/content/ventas_preparadas.csv")
print("Número de registros:", len(df))
df.head()
```

## 2. Recuperemos las medidas de posición
Antes de estudiar la dispersión, recuperemos dos medidas que ya utilizamos en el laboratorio anterior:

- **media**: valor promedio de las transacciones;
- **mediana**: valor que divide las observaciones ordenadas en dos mitades.

Estas medidas describen la **posición central** de los datos, pero no indican qué tan separados se encuentran los valores.

```python
media = df["ventas"].mean()
mediana = df["ventas"].median()
print(f"Media:   ${media:,.2f}")
print(f"Mediana: ${mediana:,.2f}")
```

### Observe

La media y la mediana nos permiten hablar del centro de la distribución.

Sin embargo, dos conjuntos de datos pueden presentar una media similar y tener comportamientos muy diferentes.

Por tanto, necesitamos responder una nueva pregunta:

> **¿Qué tan dispersos se encuentran los valores de las transacciones?**

Comenzaremos con la medida de dispersión más sencilla: **el rango**.

## 3. ¿Cuál es la amplitud de los valores?

La primera medida de dispersión que utilizaremos es el **rango**.

El rango mide la distancia entre el valor máximo y el valor mínimo observado:

$$
R = x_{\max} - x_{\min}
$$

Esta medida permite conocer la **amplitud total** que presentan los valores de una variable.

```python
minimo = df["ventas"].min()
maximo = df["ventas"].max()
rango = maximo - minimo
print(f"Valor mínimo: ${minimo:,.2f}")
print(f"Valor máximo: ${maximo:,.2f}")
print(f"Rango:        ${rango:,.2f}")
```

### Interprete el rango

El rango obtenido es:

$$
R = \$12.437.800
$$

Esto indica que existe una **gran distancia entre la transacción de menor valor y la de mayor valor**.

Responda:

1. **¿El rango parece grande en comparación con la mediana de las transacciones?**  
2. **¿Qué valor está produciendo gran parte de esta amplitud?**  

3. **¿Es suficiente conocer el rango para describir cómo se distribuyen las 4.820 transacciones?**  

### Importante

El rango permite conocer la **amplitud total** de los datos, pero solamente utiliza dos observaciones: el mínimo y el máximo.

Por tanto, no nos indica cómo se comportan los valores que se encuentran entre esos dos extremos.

## 4. ¿Qué tan alejados están los valores respecto a la media?

El rango permite conocer la amplitud total de los datos, pero utiliza únicamente dos observaciones: el mínimo y el máximo.

Para estudiar la variabilidad considerando **todas las observaciones**, utilizaremos la **desviación estándar**.

La desviación estándar mide qué tan dispersos se encuentran los valores alrededor de la media.

- Una desviación estándar **pequeña** indica que los valores tienden a encontrarse próximos a la media.
- Una desviación estándar **grande** indica que existe mayor variabilidad en los datos.

Ahora calcularemos esta medida para la variable `ventas`.

```python
desviacion = df["ventas"].std()
print(f"Media:               ${media:,.2f}")
print(f"Desviación estándar: ${desviacion:,.2f}")
```

### Interprete la desviación estándar

Los resultados obtenidos son:

- **Media:** $272.968,53
- **Desviación estándar:*
* $377.175,86

Responda:

1. **¿La desviación estándar es mayor o menor que la media?**  


2. **¿Qué indica este resultado sobre la variabilidad del valor de las transacciones?**  

3. **¿Qué diferencia existe entre la información proporcionada por el rango y la desviación estándar?**  

4. **¿Los valores elevados observados anteriormente podrían influir en la desviación estándar? ¿Por qué?**  

### Importante

La desviación estándar considera **todas las observaciones** y resume su dispersión alrededor de la media.

En este caso, su valor elevado indica una **variabilidad considerable** en el valor de las transacciones.

Sin embargo, la desviación estándar utiliza la media como referencia y puede verse afectada por **valores extremos**.

## 5. ¿Cómo se distribuye la zona central de las ventas?

Hasta ahora utilizamos el rango y la desviación estándar para estudiar la dispersión.

Ahora analizaremos la distribución mediante los **cuartiles**, que permiten localizar posiciones dentro de los datos ordenados:

- $Q_1$: aproximadamente el 25% de las observaciones se encuentra por debajo o hasta este valor.
- $Q_2$: corresponde a la mediana y representa el 50%.
- $Q_3$: aproximadamente el 75% de las observaciones se encuentra por debajo o hasta este valor.

Calcularemos estos tres puntos para la variable `ventas`.

```python
q1 = df["ventas"].quantile(0.25)
q2 = df["ventas"].quantile(0.50)
q3 = df["ventas"].quantile(0.75)
print(f"Q1 (25%): ${q1:,.2f}")
print(f"Q2 (50%): ${q2:,.2f}")
print(f"Q3 (75%): ${q3:,.2f}")
```

### Interprete los cuartiles

Los resultados obtenidos son:

- $Q_1$ = $90.350

- $Q_2$ = $166.225

- $Q_3$ = $311.820

Responda:

1. **¿Qué significa que $Q_1$ sea $90.350?**  

2. **¿Qué significa que $Q_2$ sea $166.225?**  

3. **¿Qué significa que $Q_3$ sea $311.820?**  


4. **¿Qué porcentaje aproximado de las transacciones se encuentra entre $Q_1$ y $Q_3$?**  


### Recuerde

Los cuartiles son **puntos de posición** dentro de los datos ordenados.

Entre $Q_1$ y $Q_3$ se encuentra aproximadamente el **50% central de los datos**.

Ahora podemos preguntarnos:

> **¿Qué amplitud ocupa ese 50% central de las transacciones?**

## 6. ¿Qué tan disperso está el 50% central de las ventas?

Los cuartiles nos permitieron identificar las posiciones de $Q_1$, $Q_2$ y $Q_3$.

Ahora utilizaremos $Q_1$ y $Q_3$ para calcular una medida de dispersión: el **rango intercuartílico (IQR)**.

El IQR mide la amplitud ocupada por el **50% central de las observaciones**:

$$
IQR = Q_3 - Q_1
$$

A diferencia del rango total, el IQR no depende directamente del valor mínimo y del valor máximo.

```python
iqr = q3 - q1
print(f"Q1:  ${q1:,.2f}")
print(f"Q3:  ${q3:,.2f}")
print(f"IQR: ${iqr:,.2f}")
```

### Interprete el rango intercuartílico

Los resultados obtenidos son:

- $Q_1$ = **$90.350**

- $Q_3$ = **$311.820**

- $IQR$ = **$221.470**

Responda:

1. **¿Qué representa el valor de $221.470 obtenido para el IQR?**  


2. **¿Qué porcentaje aproximado de las transacciones estamos describiendo mediante esta medida?**  

3. **¿Qué diferencia existe entre el rango total y el IQR?**


4. **¿Por qué el IQR puede resultar útil cuando existen valores extremos?**

### Importante

El **rango total** describe la amplitud entre el valor mínimo y el máximo.

El **IQR** describe la amplitud del **50% central de las observaciones**.

Por tanto, ambas son medidas de dispersión, pero describen aspectos diferentes de la distribución.

## 7. ¿Qué información aporta cada medida?

Durante el laboratorio utilizamos diferentes medidas para describir la variable `ventas`.

Obtuvimos:

- **Media:** $272.968,53

- **Mediana:** $166.225

- **Rango:** $12.437.800

- **Desviación estándar:** $377.175,86

- **Q1:** $90.350

- **Q3:** $311.820

- **IQR:** $221.470

### Interprete

1. **¿Qué medida describe la amplitud total de las ventas?**  

   El **rango**, cuyo valor es **$12.437.800**, describe la distancia entre la transacción de menor valor y la de mayor valor.

2. **¿Qué medida describe la dispersión de las observaciones alrededor de la media?**  

   La **desviación estándar**, cuyo valor es aproximadamente **$377.175,86**, resume qué tan dispersas se encuentran las transacciones alrededor de la media.

3. **¿Qué medida describe la amplitud del 50% central de las transacciones?**  

   El **rango intercuartílico (IQR)**. En este caso, el 50% central de las transacciones se encuentra entre **$90.350 y $311.820**, ocupando una amplitud de **$221.470**.

4. **¿Cuál de estas medidas es menos sensible a los valores extremos: el rango o el IQR?**  

   El **IQR** es menos sensible a los valores extremos porque utiliza $Q_1$ y $Q_3$ y no depende directamente del valor mínimo y del valor máximo.

5. **¿Por qué no sería suficiente utilizar únicamente la media para describir la variable `ventas`?**  

   Porque la media únicamente proporciona información sobre el promedio y no describe la dispersión de los datos. Además, en este conjunto existen valores elevados que pueden aumentar la media. Por esta razón, es necesario interpretarla junto con otras medidas como la mediana, la desviación estándar y el IQR.

### Idea central

Las medidas estadísticas no compiten entre sí: **describen aspectos diferentes de los datos**.

Una descripción adecuada de una variable requiere interpretar conjuntamente su **posición y dispersión**.
