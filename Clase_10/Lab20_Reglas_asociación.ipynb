## 2. Carga y reconocimiento de las compras

Cada transacción representa una compra y contiene una lista
de productos.

Usaremos las mismas seis compras de la presentación.

Dos compras diferentes pueden contener exactamente los mismos
productos. Por eso conservamos T1 y T3 como transacciones distintas.

```python
transacciones = [
    ["pan", "leche"],             # T1
    ["pan", "leche", "huevos"],   # T2
    ["pan", "leche"],             # T3
    ["pan", "huevos"],            # T4
    ["huevos"],                   # T5
    ["arroz", "huevos"]           # T6
]
tabla_compras = pd.DataFrame({
    "compra": [
        f"T{i}" for i in range(1, len(transacciones) + 1)
    ],
    "productos": [
        ", ".join(compra) for compra in transacciones
    ]
})
print("Número de compras:", len(transacciones))
display(tabla_compras)
```

## 3. Transformación de las compras

Para aplicar Apriori, construiremos una tabla donde:

- Cada fila representa una compra.
- Cada columna representa un producto.
- **1:** el producto está presente en esa compra.
- **0:** el producto está ausente.

Estos valores indican presencia o ausencia, no cantidades compradas.

Utilizaremos `TransactionEncoder` para realizar la transformación.
Internamente conservaremos valores booleanos (`True` y `False`);
mostraremos la tabla con 1 y 0 para facilitar su lectura.

```python
codificador = TransactionEncoder()
matriz = codificador.fit(transacciones).transform(transacciones)
datos = pd.DataFrame(
    matriz,
    columns=codificador.columns_,
    index=tabla_compras["compra"]
)
display(datos.astype(int))
```

## 4. Identificación de conjuntos frecuentes con Apriori

Usaremos un soporte mínimo del **50 %**.

Como tenemos seis compras, un producto o conjunto de productos
debe aparecer en **al menos tres compras** para conservarse.

Configuraremos:

- `min_support=0.50`: soporte mínimo del 50 %.
- `use_colnames=True`: mostrar los nombres de los productos.

La tabla resultante tendrá dos columnas:

- **support:** proporción de compras que contiene el conjunto.
- **itemsets:** productos que forman el conjunto.

En este paso obtenemos conjuntos frecuentes.
Las reglas con antecedente y consecuente se generarán después.

```python
frecuentes = apriori(
    datos,
    min_support=0.50,
    use_colnames=True
)
print("Número de conjuntos frecuentes:", len(frecuentes))
display(frecuentes)
```

## 5. Generación de reglas de asociación

Usaremos una confianza mínima del 70 %.

Una regla A → B se conservará si, entre las compras que contienen
A, al menos el 70 % también contiene B.

La función `association_rules` genera las reglas a partir de
los conjuntos frecuentes y calcula sus métricas:

- **Soporte:** proporción de todas las compras que contienen A y B.
- **Confianza:** proporción de compras con B entre las que contienen A.
- **Lift:** compara la confianza con la frecuencia general de B.

Los conjuntos frecuentes ya cumplen el soporte mínimo del 50 %.
Ahora filtraremos las reglas por su confianza.

```python
reglas = association_rules(
    frecuentes,
    num_itemsets=len(datos),
    metric="confidence",
    min_threshold=0.70
)
columnas = [
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]
print("Número de reglas encontradas:", len(reglas))
display(reglas[columnas].round(3))
```

## 6. ¿Qué ocurre al aumentar la confianza mínima?

Antes conservamos las reglas con confianza igual o superior al 70 %.
Ahora aumentaremos ese mínimo al 90 %, manteniendo el soporte mínimo
del 50 %.

Antes de ejecutar: ¿cuál de las dos reglas esperas conservar?

```python
reglas_90 = association_rules(
    frecuentes,
    num_itemsets=len(datos),
    metric="confidence",
    min_threshold=0.90
)
print("Reglas con confianza mínima del 70 %:", len(reglas))
print("Reglas con confianza mínima del 90 %:", len(reglas_90))
display(reglas_90[columnas].round(3))
```
