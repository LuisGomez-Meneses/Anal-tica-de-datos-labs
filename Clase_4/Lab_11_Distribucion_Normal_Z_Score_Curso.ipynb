## 1. Carga de la base de datos

que fue simulada a partir de una distribución Normal.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

datos = pd.read_csv("/content/datos_distribuciones.csv")

datos.head()
```

## 2. Media y desviación estándar

Antes de aplicar la regla 68–95–99.7 necesitamos conocer el centro
y la dispersión de la variable `peso_producto`.

Calcularemos nuevamente su media y su desviación estándar.

```python
media = datos["peso_producto"].mean()
desviacion = datos["peso_producto"].std()
print(f"Media: {media:.2f} g")
print(f"Desviación estándar: {desviacion:.2f} g")
```

```python
limite_inferior_1 = media - desviacion
limite_superior_1 = media + desviacion
print(f"Límite inferior: {limite_inferior_1:.2f} g")
print(f"Límite superior: {limite_superior_1:.2f} g")
```

## 4. Comprobación de la regla del 68 %

Según la regla 68–95–99.7, en una distribución aproximadamente normal
esperamos que cerca del **68 % de los valores** se encuentre entre:

$
\mu-\sigma \leq X \leq \mu+\sigma
$

Para nuestra variable `peso_producto`, este intervalo es aproximadamente:

$
473.68 \leq X \leq 523.66
$

Vamos a comprobar qué porcentaje de las observaciones de nuestra base
se encuentra realmente dentro de este intervalo.

```python
dentro_1_sigma = datos[
    (datos["peso_producto"] >= limite_inferior_1) &
    (datos["peso_producto"] <= limite_superior_1)
]
cantidad_1_sigma = len(dentro_1_sigma)
porcentaje_1_sigma = (
    cantidad_1_sigma / len(datos)
) * 100
print(f"Observaciones dentro de ±1σ: {cantidad_1_sigma}")
print(f"Porcentaje observado: {porcentaje_1_sigma:.2f}%")
print("Porcentaje teórico aproximado: 68%")
```

```python
limite_inferior_2 = media - 2 * desviacion
limite_superior_2 = media + 2 * desviacion

dentro_2_sigma = datos[
    (datos["peso_producto"] >= limite_inferior_2) &
    (datos["peso_producto"] <= limite_superior_2)
]

cantidad_2_sigma = len(dentro_2_sigma)
porcentaje_2_sigma = (cantidad_2_sigma / len(datos)) * 100

print(f"Intervalo: {limite_inferior_2:.2f} g a {limite_superior_2:.2f} g")
print(f"Observaciones dentro de ±2σ: {cantidad_2_sigma}")
print(f"Porcentaje observado: {porcentaje_2_sigma:.2f}%")
print("Porcentaje teórico aproximado: 95%")
```

## 6. Comprobación de la regla del 99.7 %

Finalmente, ampliaremos el intervalo a **tres desviaciones estándar**
a cada lado de la media:

\[
\mu-3\sigma \leq X \leq \mu+3\sigma
\]

Según la regla 68–95–99.7, aproximadamente el **99.7 % de los valores**
de una distribución Normal se encuentra dentro de este intervalo.

Comprobaremos qué porcentaje obtenemos en nuestra base de datos.

```python
limite_inferior_3 = media - 3 * desviacion
limite_superior_3 = media + 3 * desviacion
dentro_3_sigma = datos[
    (datos["peso_producto"] >= limite_inferior_3) &
    (datos["peso_producto"] <= limite_superior_3)
]
cantidad_3_sigma = len(dentro_3_sigma)
porcentaje_3_sigma = (cantidad_3_sigma / len(datos)) * 100
print(f"Intervalo: {limite_inferior_3:.2f} g a {limite_superior_3:.2f} g")
print(f"Observaciones dentro de ±3σ: {cantidad_3_sigma}")
print(f"Porcentaje observado: {porcentaje_3_sigma:.2f}%")
print("Porcentaje teórico aproximado: 99.7%")
```

## 8. Cálculo de la puntuación Z

Hasta ahora analizamos intervalos alrededor de la media.

Ahora queremos determinar la posición de **cada observación individual**
respecto a la media utilizando la puntuación Z.

Recordemos:

$
Z=\frac{x-\mu}{\sigma}
$

El valor de Z indica cuántas desviaciones estándar se encuentra una
observación por encima o por debajo de la media.

```python
datos["z_score"] = (
    (datos["peso_producto"] - media) / desviacion
)
datos[["peso_producto", "z_score"]].head()
```

## 9. Observaciones más alejadas de la media

La magnitud del Z-score permite determinar qué tan lejos se encuentra
una observación de la media.

- Un Z positivo indica un valor por encima de la media.
- Un Z negativo indica un valor por debajo de la media.
- Cuanto mayor sea $|Z|$, mayor será la distancia respecto a la media.

Buscaremos los productos con el Z-score más bajo y más alto de nuestra base.

```python
indice_min = datos["z_score"].idxmin()
indice_max = datos["z_score"].idxmax()
observacion_min = datos.loc[
    indice_min,
    ["id", "peso_producto", "z_score"]
]
observacion_max = datos.loc[
    indice_max,
    ["id", "peso_producto", "z_score"]
]
print("Observación con menor Z-score:")
print(observacion_min)
print("\nObservación con mayor Z-score:")
print(observacion_max)
```
